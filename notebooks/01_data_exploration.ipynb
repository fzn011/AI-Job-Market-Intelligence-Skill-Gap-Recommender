# 01 — Data Exploration

**Purpose:** Explore the sample job postings dataset.  
Get a feel for data shape, distributions, missing values, and initial text patterns.

---

```python
# Standard imports
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import plotly.express as px
from pathlib import Path

print('Libraries loaded.')
```

```python
# Load the sample jobs dataset
DATA_PATH = Path('..') / 'data' / 'sample' / 'sample_jobs.csv'

df = pd.read_csv(DATA_PATH)
print(f'Loaded {len(df)} job postings.')
df.head()
```

```python
# Basic info
print('Shape:', df.shape)
print('\nColumn names:', df.columns.tolist())
print('\nData types:')
print(df.dtypes)
```

```python
# Check for missing values
print('Missing values per column:')
print(df.isnull().sum())
```

```python
# Distribution of job titles
title_counts = df['job_title'].value_counts().reset_index()
title_counts.columns = ['job_title', 'count']

fig = px.bar(
    title_counts,
    x='job_title',
    y='count',
    title='Job Title Distribution',
    labels={'job_title': 'Job Title', 'count': 'Count'},
    color='count',
    color_continuous_scale='Blues'
)
fig.update_layout(xaxis_tickangle=-30)
fig.show()
```

```python
# Distribution of job types
print('Job types:\n', df['job_type'].value_counts())
print('\nLocations:\n', df['location'].value_counts())
```

```python
# Description length analysis
df['desc_length'] = df['description'].apply(lambda x: len(str(x).split()))

print('Description word count stats:')
print(df['desc_length'].describe())

fig2 = px.histogram(
    df,
    x='desc_length',
    nbins=20,
    title='Distribution of Description Word Counts',
    labels={'desc_length': 'Word Count'}
)
fig2.show()
```
